```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

bac = pd.read_csv("fr-en-baccalaureat-par-academie.csv",sep=";")
bac = bac.loc[bac["Voie"] == "BAC GENERAL"]

bac["Sexe"]= bac["Sexe"].replace("MASCULIN", "GARCONS")
bac["Sexe"]= bac["Sexe"].replace("FEMININ", "FILLES")
bac.head(5)

bac.groupby("Libellé académie").agg({"Nombre d'inscrits": 'sum',"Nombre de présents":"sum","Nombre d'admis totaux": 'sum',"Nombre de refusés totaux":"sum"})
```
