Nous avons cherché à relier le profil des néo-bacheliers admis dans chaque département (Parcoursup 2023 : part de bacheliers généraux, technologiques et professionnels, et taux de mention) à la situation économique locale. Un premier test avec la part des actifs ayant un emploi (recensement 2023) n'a montré aucun lien (corrélations proches de 0). Cette variable, qui mesure en réalité l'envers du chômage et non le taux d'activité, varie très peu d'un département à l'autre et ne suit pas le gradient de dynamisme économique. En la remplaçant par le niveau de vie médian des ménages (Filosofi 2023), on observe un lien net : environ +0,5 avec la part de bacheliers généraux et -0,5 avec celles des bacheliers technologiques et professionnels.

Ces résultats restent des associations et non des relations causales. Aussi, elles reflètent en partie l'urbanisation, car les grandes villes concentrent à la fois les formations sélectives, qui recrutent surtout des bacheliers généraux, et les hauts revenus. De plus, Parcoursup localise les établissements d'accueil et non les habitants, alors que Filosofi mesure les ménages résidents. Une prochaine étape qui aurait pu être intéressante avec plus de temps aurait été d'ajouter des variables de contrôle (densité, part de grandes villes) et un indicateur de diplôme de la population issu du recensement.
