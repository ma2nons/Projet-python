# IV. Limite des résultats

Nous avons dans la partie précédente mis en lumière une corrélation entre le taux de réussite au baccalauréat et le taux d'activité dans un même département. Néanmoins, nous pouvons interroger la solidité des corrélations obtenues précédemment, et surtout questionner une éventuelle causalité du taux d'activité sur la variation du taux de réussite au baccalauréat.

## IV. a. Robustesse des résultats dans le temps

Pour les mettre à l'épreuve, nous avons choisi de réaliser une analyse diachronique, en nous appuyant sur trois coupes transversales (2012,2017 et 2023).

Dans un premier temps, nous nous sommes intéressés à la période la plus récente, à savoir 2017-2023 (Pour le choix des dates, se référer à la section IV.c). Nous avons obtenu ces résultats:


```python
display(Image(filename="output/graphique_progression_2017_3.png"))
```

Nous avons étudié dans un premier temps l'éventuelle corrélation entre une progression du taux d'activité de 2023 par rapport à 2017 pour un département donné, et une progression aux mêmes dates de son taux de réussite au baccalauréat (résultats exprimés ici en points de pourcentage. Lecture: un point indiquant une progression de 0.05 en taux d'activité signifie une augmentation de 5 points de pourcentage du taux d'activité entre 2017 et 2023. Par exemple, le taux d'activité passe de 91% à 96%). 

La figure ci-contre obtenue ne plaide pas en faveur d'une corrélation, les points étant tous plus ou moins regroupés autour d'une progression de taux de réussite de l'ordre de 4%, et ce quel que soit la progression du taux d'activité. Forts de ce résultat, nous avons voulu visualiser plus précisément les données de 2017 et 2023, notamment pour tester l'hypothèse selon laquelle les progressions ne seraient pas corrélées car elles ne rendraient pas compte de l'information concernant les taux de réussite et d'activité initiaux des départements (exemple: un faible taux de progression du taux de réussite scolaire et du taux d'activité pour un département chez qui ces taux sont déjà élevés initialement n'a pas la même signification que pour un département chez qui ces taux sont plus faibles. )

```python
display(Image(filename="output/graphique_nuage_activite_17_3.png"))
display(Image(filename="output/graphique_fleches_2017_3.png"))
```

Ces deux figures nous ont permis de confirmer notre intuition : presque tous les taux d'activité et tous les taux de réussite ont évolué favorablement entre 2017 et 2023. 
Nous l'observons d'abord sur le premier graphique représentant la relation entre taux d'activité et taux de réussite au baccalauréat, où les points bleus sont bien plus resserés que les points rouges, montrant une tendance nationale à la hausse des taux de réussite au baccalauréat et des taux d'activité.
Nous le voyons de façon encore plus nette dans le deuxième graphique, qui pour chaque département associe son point initial de 2017 (queue de la flèche) à son point final en 2023 (tête de la flèche). Le fait que toutes les flèches ou presque soient vertes (ie une progression ET du taux de réussite au baccalauréat, ET du taux d'activité) nous a semblé surprenant, et nous a incité à la prudence quant à l'interprétation de nos résultats.

En effet, ou bien il existait une corrélation extrêmement forte entre l'évolution du taux d'activité, et l'évolution du taux de réussite au baccalauréat, ou bien nous étions sujets au biais de la variable omise. Le fait qu'il y ait peu de flèches jaunes (les flèches étant jaunes lorsque le taux d'activité et le taux de réussite au baccalauréat n'évoluent pas dans le même sens) peut nous inciter à penser qu'il y a bien une corrélation forte, mais l'existence d'aucune flèche rouge (la flèche est rouge si ET le taux de réussite au baccalauréat, ET le taux d'activité diminuent)nous laissait sceptiques: la corrélation semblait ne pas pouvoir se transformer en causalité, et être plus liée à des facteurs conjoncturels que structurels (voir section IV.b). 
Nous avons donc décidé de réitérer ces analyses sur la période précédente (2012-2017), en voulant tester la robustesse de nos corrélations et l'hypothèse de causalité dans un contexte de conjoncture différente.

Cette figure  promet des résultats intéressants pour les autres graphiques car les progression de taux d'activité n'évoluent pas tous favorablement. Ainsi, le contexte conjoncturel économique semble différent de celui de 2017 et de 2023, et se prêtera mieux à l'analyse de la corrélation notamment pour le graphique en flèches (forte présence de flèches vertes et rouges, renforçant l'hypothèse d'une causalité ou forte présence de flèches jaunes, l'affaiblissant).

## IV.b. Hypothèses sur les facteurs entrant en compte

Après avoir observé les résultats précédents, nous avons voulu identifier les différents paramètres qui pourraient expliquer de tels résultas dans notre analyse diachronique. 
Nous avons identifié comme causes potentielles de l'augmentation de la réussite au baccalauréat qui ne serait pas liée à l'augmentation du taux d'activité les facteurs suivants:
- la réforme de la voie professionnelle entre 2011 et 2014, conduisant à une augmentation du taux de réussite au baccalauréat professionnel (DEPP n°17.05, "Résultats définitifs de la session 2016 du baccalauréat")
- la création d'un nouveau baccalauréat dont la première session fut créée en 2021 - baccalauréat accordant une place plus importante au contrôle continu que sous son ancien format


## IV. c. Limites inhérentes aux données

Notre travail nous a également conduit à identifier des limites inhérentes aux bases de données et/ou aux données étudiées. 

Concernant notre travail sur les données parcoursup, nous regrettons l'absence d'information sur les trajectoires individuelles de mobilité des bacheliers, indisponibles pour des raisons de confidentialité et de vie privée. Ainsi, ne disposant que d'une approximation du taux de mobilité à l'échelle agrégée des départements, la portée de nos résultats et interprétations s'en voit nécessairement diminuée. La question de la mobilité est pourtant centrale. Si les bacheliers changent de département, il n'est plus possible d'évaluer l'effet du résultat scolaire dans un département sur l'emploi car les personnes qui y travaillent alors ne sont plus les mêmes que celles qui y ont étudié. C'est la raison pour laquelle, entre autres, il nous était plus difficile de travailler sur l'effet de l'éducation sur l'emploi que sur l'effet de l'emploi sur l'éducation. En outre, l'effet de l'éducation sur l'emploi ne se voyant qu'à moyen terme, au regard des données dont nous disposions (à l'espace temporel assez réduit) il semblait difficile d'étudier de façon satisfaisante cet effet. Pour ces raisons nous avons donc fait le choix de nous restreindre à l'effet de l'emploi sur l'éducation.

Concernant notre travail sur l'analyse diachronique, nous regrettons d'avoir eu accès aux données d'activité uniquement pour les années 2012,2017 et 2023, nous empêchant de produire des résultats chaînés qui auraient été bien plus robustes. N'utiliser que les années 2012,2017 et 2023 nous oblige à supposer qu'elles ne sont pas des années particulières, mais plutôt des années représentatives de la période 2012-2017 et 2017-2023. Si nous avions les données de taux de réussite au baccalauréat pour toutes les années, nous avons néanmoins préféré ne pas réaliser d'indicateur chaîné au niveau scolaire, dans un soucis de cohérence et de comparabilité avec l'indicateur économique. En outre, afin d'affiner nos résultats, nous aurions voulu réaliser la même analyse que pour les taux de réussite au baccalauréat pour les taux de mentions - indicateur qui, comme le montrait la section III, aurait permis de mieux affiner les résultats -  mais l'absence de données sur les mentions en 2012 nous en a empêché.

De manière générale, en étudiant les données d'éducations, nous avons fait face à la difficulté suivante : la plupart sont produites par académie, alors que celles sur l'activité sont produites au niveau départementale. Il est alors plus difficile de comparer les deux. Parfois, même l'étude des données d'éducation seules, sans essayer de les comparer aux autres variables de contrôles, à l'échelle départementale, est impossible. Ce fut le cas notamment pour contrôler la mobilité des bacheliers en sortant du lycée.
